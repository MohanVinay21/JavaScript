# 2. Strings and String Literals

## 2.1 What is a string?

A string is a sequence of text characters represented in JavaScript using UTF-16 code units.

Strings are used to represent names, messages, addresses, sentences, and other textual data.

```javascript
let name = "Vinay";
let message = "JavaScript is interesting";
```

## 2.2 What is a string literal?

A string literal is the text written directly in source code to create a string value.

JavaScript supports three common ways to write string literals.

**A. Single quotes**

```javascript
let language = 'JavaScript';
```

**B. Double quotes**

```javascript
let language = "JavaScript";
```

Both single-quoted and double-quoted strings behave similarly.

**C. Template literals**

Template literals use backticks rather than ordinary quotation marks.

```javascript
let language = `JavaScript`;
```

Template literals support string interpolation and multiline text.

## 2.3 Escape sequences

Escape sequences allow special characters to be represented inside string literals.

| **Escape sequence** | **Meaning** |
|---|---|
| `\n` | New line |
| `\t` | Horizontal tab |
| `\\` | Backslash |
| `\'` | Single quote |
| `\"` | Double quote |
| `` ` `` | Backtick inside a template literal |

**Example 1: New line**

```javascript
let message = "Hello\nWorld";
console.log(message);
```

**Output:**

```text
Hello
World
```

**Example 2: Apostrophe inside a string**

```javascript
let sentence = "It's a beautiful day.";
console.log(sentence);
```

**Output:**

```text
It's a beautiful day.
```

The apostrophe does not need escaping because the string is enclosed in double quotes.

## 2.4 Template literals and interpolation

Template literals allow expressions to be embedded inside a string using `${...}`.

**Example 1: Variable interpolation**

```javascript
let name = "Vinay";
let age = 19;

let message = `My name is ${name} and I am ${age} years old.`;

console.log(message);
```

**Output:**

```text
My name is Vinay and I am 19 years old.
```

**Example 2: Expressions inside interpolation**

Expressions can also be evaluated inside the interpolation.

```javascript
let a = 10;
let b = 20;

console.log(`The sum is ${a + b}.`);
```

**Output:**

```text
The sum is 30.
```

**Example 3: Multiline template literals**

Template literals can span multiple lines.

```javascript
let receipt = `Product: Keyboard
Price: 1200
Quantity: 2`;

console.log(receipt);
```

This is useful when creating receipts, formatted messages, reports, and multiline text.

## 2.5 String immutability

JavaScript strings are **immutable**, meaning their individual characters cannot be changed in place.

```javascript
let text = "Hello";

// text[0] = "Y"; // Does not change the string

console.log(text); // Hello
```

To create a modified string, assign a new string value:

```javascript
let text = "Hello";
text = "Y" + text.slice(1);

console.log(text); // Yello
```

The variable can be reassigned because it was declared with `let`, but the original string itself was not modified.

## 2.6 String length and character access

The `.length` property returns the number of UTF-16 code units in a string.

**Example 1: String length**

```javascript
let message = "JavaScript";

console.log(message.length); // 10
```

**Example 2: Character access**

Characters can be accessed using zero-based indexing.

```javascript
let word = "Hello";

console.log(word[0]); // H
console.log(word[1]); // e
console.log(word[word.length - 1]); // o
```

The first character is at index `0`. The last character is at index `length - 1`.

**Advanced detail: Unicode characters**

`.length` does not always equal the number of user-perceived characters. Some Unicode symbols, including many emoji, use multiple UTF-16 code units.

```javascript
console.log("😀".length); // 2
```

## 2.7 Important string methods

### A. Case conversion

```javascript
let text = "JavaScript";

console.log(text.toUpperCase()); // JAVASCRIPT
console.log(text.toLowerCase()); // javascript
```

These methods return new strings; they do not mutate the original string.

### B. Removing whitespace

```javascript
let text = "   Hello World   ";

console.log(text.trim());      // "Hello World"
console.log(text.trimStart()); // Removes leading whitespace
console.log(text.trimEnd());   // Removes trailing whitespace
```

### C. Searching within a string

```javascript
let text = "I love JavaScript";

console.log(text.includes("JavaScript")); // true
console.log(text.indexOf("love"));        // 2
console.log(text.indexOf("Python"));      // -1
```

- `includes()` checks whether a substring exists.
- `indexOf()` returns the first matching index or `-1` if no match exists.

These searches are case-sensitive.

### D. Extracting a substring

```javascript
let text = "JavaScript";

console.log(text.slice(0, 4)); // Java
console.log(text.slice(4));    // Script
console.log(text.slice(-6));   // Script
```

The ending index in `slice(start, end)` is excluded.

### E. Replacing text

```javascript
let text = "I love Python and Python";

console.log(text.replace("Python", "JavaScript"));
// I love JavaScript and Python
```

For an ordinary string search, `replace()` replaces only the first occurrence.

**Replacing all occurrences:**

```javascript
let text = "Python is fun. Python is powerful.";

console.log(text.replaceAll("Python", "JavaScript"));
// JavaScript is fun. JavaScript is powerful.
```

### F. Splitting and joining

```javascript
let fruits = "apple,banana,mango";

let array = fruits.split(",");

console.log(array);
// ["apple", "banana", "mango"]

console.log(array.join(" - "));
// apple - banana - mango
```

- `split()` divides a string into an array.
- `join()` combines array elements into a string.

## 2.8 String concatenation

Concatenation means joining strings together.

**A. Using the `+` operator**

```javascript
let firstName = "Mohan";
let lastName = "Vinay";

let fullName = firstName + " " + lastName;

console.log(fullName); // Mohan Vinay
```

**B. Using a template literal**

```javascript
let firstName = "Mohan";
let lastName = "Vinay";

let fullName = `${firstName} ${lastName}`;

console.log(fullName); // Mohan Vinay
```

For combining variables and expressions, template literals are often easier to read.

## 2.9 Type conversion and string coercion

JavaScript sometimes converts values automatically when an operation requires a different type. This is called **type coercion**.

**Example 1: Automatic conversion**

```javascript
console.log("5" + 2); // "52"
console.log("5" - 2); // 3
```

In the first example, `+` performs string concatenation because one operand is a string.

In the second example, subtraction converts the string `"5"` to the number `5`.

**Example 2: Explicit conversion**

```javascript
let value = "25";

console.log(Number(value) + 5); // 30
console.log(String(100));       // "100"
```

**Example 3: Invalid numeric strings**

```javascript
console.log(Number("hello")); // NaN
```

`NaN` means "Not a Number" and represents an invalid numeric result.

## 2.10 Comparing strings

Strings can be compared using equality operators.

```javascript
console.log("hello" === "hello"); // true
console.log("hello" === "Hello"); // false
```

String comparisons are case-sensitive.

Prefer strict equality (`===`) because it compares values without performing the type coercion associated with loose equality (`==`).

```javascript
console.log(5 == "5");  // true
console.log(5 === "5"); // false
```

# 3. Common Mistakes to Avoid

- Using `const` and then trying to reassign the variable.
- Assuming `const` makes arrays and objects completely immutable.
- Accessing a `let` or `const` variable before its declaration is evaluated.
- Forgetting that JavaScript is case-sensitive.
- Confusing `undefined` with `null`.
- Using ordinary quotes when template literal interpolation is needed.
- Forgetting that string indexing starts at `0`.
- Assuming `replace()` replaces every occurrence.
- Assuming `+` always performs numeric addition.
- Trying to modify a string character directly.

# 4. Best Practices

- Use `const` unless reassignment is necessary.
- Use `let` when the variable must be reassigned.
- Avoid `var` in modern JavaScript unless you specifically need its behavior.
- Choose meaningful variable names and use camelCase.
- Prefer template literals for readable strings containing variables.
- Use strict equality (`===`) by default.
- Keep string transformations readable and avoid unnecessarily complicated expressions.
- Test edge cases, including empty strings, missing values, and unexpected input types.
