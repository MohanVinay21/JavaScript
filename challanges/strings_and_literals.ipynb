```typescript
//converting the names to lowercase and combining them using template literals
const firstname = 'Mohan';
const lastname = 'Vinay';
const dob = 2007;
const username = `${firstname.toLowerCase()}.${lastname.toLowerCase()}${dob}`;
console.log(username);
```

```typescript
let product = "Keyboard";
let price = 1200;
let quantity = 2;
let discount = 10;

let price_before_discount = price * quantity;
let discount_amount = (price_before_discount * discount) / 100;
let final_amount = price_before_discount - discount_amount;

const receipt = 
`================reciept=======================
ProductName = ${product}
Price Per Item = ${price}
Total Quantity = ${quantity}
Total Before Discount = ${price_before_discount}
Discount Amount = ${discount_amount}
-------------------------------------------------
Final Amount = ${final_amount}
================================================`;
console.log(receipt);
```
