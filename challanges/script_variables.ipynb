# JavaScript Notes — Variables

## 1. Variables in JavaScript

### 1.1 What is a variable?

A variable is a named reference used to store or access a value in a program. In JavaScript, variables can hold different types of values, including numbers, strings, booleans, objects, and functions.

```javascript
let age = 19;
let name = "Vinay";
let isStudent = true;
```

JavaScript is a dynamically typed language. This means you do not need to specify a variable's data type when declaring it. Its type is determined by the value assigned to it, and a variable declared with `let` can later hold a value of another type.

```javascript
let value = 100;
value = "Hello";
value = false;
```

### 1.2 Ways to declare variables

JavaScript provides three main variable declaration keywords: `var`, `let`, and `const`.

#### A. let

- Used when a variable's value needs to change.
- Has block scope.
- Cannot be redeclared in the same scope.
- Can be reassigned.

```javascript
let score = 50;
score = 80;

console.log(score); // 80
```

#### B. const

- Used when a variable should not be reassigned.
- Has block scope.
- Must be initialized when declared.
- Cannot be redeclared in the same scope.

```javascript
const pi = 3.14159;
console.log(pi);

// pi = 3.14; // TypeError
```

**Important:** `const` prevents reassignment of the variable binding. It does not automatically make an object or array immutable.

```javascript
const student = {
    name: "Vinay",
    age: 19
};

student.age = 20; // Allowed
console.log(student.age); // 20

// student = {}; // TypeError
```

The object can be modified, but the variable `student` cannot be reassigned to a different object.

#### C. var

- Has function scope rather than block scope.
- Can be redeclared and reassigned.
- Can be accessed before its declaration in certain situations because it is hoisted and initialized with `undefined`.
- Is generally avoided in modern JavaScript when `let` or `const` is more appropriate.

```javascript
var x = 10;
var x = 20;

console.log(x); // 20
```

### 1.3 Difference between var, let, and const

| Feature | var | let | const |
|---|---|---|---|
| Scope | Function | Block | Block |
| Reassignment | Allowed | Allowed | Not allowed |
| Redeclaration in the same scope | Allowed | Not allowed | Not allowed |
| Must initialize immediately | No | No | Yes |
| Hoisting | Yes, initialized to `undefined` | Yes, inaccessible before declaration | Yes, inaccessible before declaration |

**Best practice:** Prefer `const` by default. Use `let` when reassignment is needed. Avoid `var` in most modern JavaScript.

### 1.4 Scope of variables

Scope determines where a variable can be accessed in a program.

#### A. Global scope

A variable declared outside functions and blocks may be accessible throughout its applicable scope.

```javascript
let message = "Hello";

function greet() {
    console.log(message);
}

greet(); // Hello
```

#### B. Block scope

A block is a section of code enclosed in curly braces `{}`. Variables declared with `let` and `const` inside a block are accessible only within that block.

```javascript
if (true) {
    let score = 90;
    const grade = "A";

    console.log(score); // 90
}

// console.log(score); // ReferenceError
```

#### C. Function scope

A variable declared with `var` inside a function is accessible throughout that function, including outside nested blocks.

```javascript
function example() {
    var x = 10;

    if (true) {
        var y = 20;
    }

    console.log(x); // 10
    console.log(y); // 20
}

example();
```

Here, `y` is accessible outside the `if` block because `var` is not block-scoped.

### 1.5 Hoisting and the Temporal Dead Zone (TDZ)

Hoisting describes JavaScript's behavior of processing declarations before executing the code in their scope.

#### A. Hoisting with var

With `var`, the declaration is hoisted and initialized to `undefined`.

```javascript
console.log(x); // undefined
var x = 10;
```

Conceptually, this behaves similarly to:

```javascript
var x;
console.log(x);
x = 10;
```

This is a conceptual illustration, not a literal transformation performed on the source code.

#### B. Hoisting with let and const

With `let` and `const`, declarations are also hoisted, but they cannot be accessed before their declaration is evaluated.

The period between entering the scope and evaluating the declaration is called the **Temporal Dead Zone (TDZ)**.

```javascript
// console.log(score); // ReferenceError
let score = 50;
```

This distinction is important because accessing a `let` or `const` variable too early causes an error instead of returning `undefined`.

### 1.6 Variable naming rules

**Valid variable names:**

```javascript
let studentName = "Vinay";
let _count = 10;
let $price = 500;
let total2 = 100;
```

**Rules:**

- Names can contain letters, digits, underscores, and dollar signs.
- Names cannot begin with a digit.
- Reserved keywords such as `if`, `class`, and `return` cannot be used as ordinary variable names.
- JavaScript is case-sensitive: `score` and `Score` are different variables.
- Use meaningful names that describe the value.
- Prefer `totalPrice` over `tp`.
- JavaScript commonly uses camelCase for variable names.

### 1.7 Data types and the typeof operator

The `typeof` operator returns a string describing the type of a value.

```javascript
let n = 25;
let s = "Hello";
let b = false;
let u;

console.log(typeof n); // "number"
console.log(typeof s); // "string"
console.log(typeof b); // "boolean"
console.log(typeof u); // "undefined"
```

**Important cases:**

```javascript
console.log(typeof 10);             // "number"
console.log(typeof 10n);            // "bigint"
console.log(typeof Symbol());       // "symbol"
console.log(typeof {});             // "object"
console.log(typeof null);           // "object" (historical quirk)
console.log(typeof function() {});  // "function"
```

**Important:** `typeof null` returning `"object"` is a long-standing JavaScript behavior. It does not mean that `null` is actually an object.

#### Understanding undefined and null

```javascript
let a;
let b = null;

console.log(a); // undefined
console.log(b); // null
```

- **undefined:** Commonly indicates that a value has not been assigned.
- **null:** An intentional value representing the absence of a value.
