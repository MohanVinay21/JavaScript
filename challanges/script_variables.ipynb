```typescript
//swaping of variables
let a = 10;
let b = 20;
let c;
console.log("before swapping");
console.log("a = ",a ,"b = ",b)
c = a;
a = b;
b = c;
console.log("after swapping");
console.log("a = ",a,"b = ",b);
```

```typescript
//swapping variables with out creating new variables
let num1 = 10;
let num2 = 20;
console.log("befre swapping the numbers");
console.log("num1 = ",num1,"num2 = ",num2);
[num1, num2] = [num2, num1];
console.log("after swapping the numbers");
console.log("num1 = ",num1,"num2 = ",num2);
```

Here I am using array destructuring, to swap variables with out using declaring the another variables
