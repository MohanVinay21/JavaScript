```typescript
//To check given number is positive or negative or zero
const num = -8;
if (num >= 0)
{
    console.log(`${num} is positive number`);
}
else if (num <= 0)
{
    console.log(`${num} is negative number`);
}
else
{
    console.log(`${num} is zero`);
}
```
