# JavaScript Notes — Conditional Statements (Advanced)

## 1. What is a conditional statement?

A conditional statement controls the flow of a program by executing different blocks of code depending on whether a condition evaluates to `true` or `false`.

```
let age = 20;

if (age >= 18) {
    console.log("You are an adult");
} else {
    console.log("You are a minor");
}
```

JavaScript evaluates the condition inside `()` and converts its result to a Boolean for the decision.

## 2. Types of conditional statements

1\. `if`

Executes a block only when the condition is truthy.

```
if (temperature > 30) {
    console.log("It's hot");
}
```

2\. `if-else`

Chooses between two paths.

```
if (isLoggedIn) {
    console.log("Welcome");
} else {
    console.log("Please log in");
}
```

3\. `if-else if-else`

Checks multiple conditions in order. The first matching branch runs.

```
let marks = 85;

if (marks >= 90) {
    console.log("A");
} else if (marks >= 80) {
    console.log("B");
} else if (marks >= 40) {
    console.log("Pass");
} else {
    console.log("Fail");
}
```

4\. Nested `if`

An `if` statement inside another conditional block.

```
let age = 22;
let hasID = true;

if (age >= 18) {
    if (hasID) {
        console.log("Entry allowed");
    } else {
        console.log("ID required");
    }
} else {
    console.log("Underage");
}
```

## 3. Comparison operators

These operators are commonly used to create conditions.

| Operator | Meaning               | Example              |
| -------- | --------------------- | -------------------- |
| `===`    | Strict equality       | `5 === 5` → `true`   |
| `!==`    | Strict inequality     | `5 !== "5"` → `true` |
| `>`      | Greater than          | `10 > 5` → `true`    |
| `<`      | Less than             | `3 < 2` → `false`    |
| `>=`     | Greater than or equal | `18 >= 18` → `true`  |
| `<=`     | Less than or equal    | `10 <= 9` → `false`  |

Advanced tip: Prefer `===` and `!==` over `==` and `!=` to avoid unexpected type conversion.

```
console.log(5 == "5");  // true
console.log(5 === "5"); // false
```

The first comparison converts values to compatible types. The second checks both type and value.

## 4. Logical operators in conditions

Logical operators combine or reverse conditions.

\| Operator | Meaning | Example | |---|---|---| | `&&` | AND — both conditions must be truthy | `age >= 18 && hasID` | | `||` | OR — at least one condition must be truthy | `isAdmin || isOwner` | | `!` | NOT — reverses truthiness | `!isLoggedIn` |

### AND (`&&`)

```
let age = 21;
let hasID = true;

if (age >= 18 && hasID) {
    console.log("Entry allowed");
}
```

Both conditions must be true.

### OR (`||`)

```
let isWeekend = false;
let isHoliday = true;

if (isWeekend || isHoliday) {
    console.log("No college today");
}
```

Only one condition needs to be true.

### NOT (`!`)

```
let isLoggedIn = false;

if (!isLoggedIn) {
    console.log("Please log in");
}
```

Because `isLoggedIn` is false, `!isLoggedIn` is true.

### Combining operators

```
let age = 20;
let hasID = true;
let isBanned = false;

if (age >= 18 && hasID && !isBanned) {
    console.log("Access granted");
} else {
    console.log("Access denied");
}
```

This is useful when several requirements must be satisfied simultaneously.

## 5. Truthy and falsy values

JavaScript does not require every `if` condition to be explicitly `true` or `false`. It converts the condition to a Boolean.

### Falsy values

These values evaluate to false in a condition:

- `false`
- `0` and `-0`
- `""` (empty string)
- `null`
- `undefined`
- `NaN`
- `0n` (BigInt zero)

Most other values are truthy, including empty arrays `[]`, empty objects `{}`, and the string `"false"`.

```
if ("false") {
    console.log("This executes");
}

if ([]) {
    console.log("This also executes");
}

if (0) {
    console.log("This does not execute");
}
```

Important: An empty array is truthy in JavaScript. Don't confuse JavaScript with languages where an empty collection may be falsy.

### Checking for a value correctly

```
let username = "";

if (username) {
    console.log("Username provided");
} else {
    console.log("Username missing");
}
```

This works if an empty string means “missing.” But if `0` is a valid value, checking truthiness could incorrectly reject it.

For example:

```
let quantity = 0;

if (quantity !== null && quantity !== undefined) {
    console.log("Quantity was provided");
}
```

This checks whether the quantity is absent without treating zero as missing.

## 6. Short-circuit evaluation

This is an important advanced concept. JavaScript may stop evaluating a logical expression as soon as its result is known.

### With `&&`

If the left side is falsy, the right side is not evaluated.

```
let age = 16;

if (age >= 18 && console.log("Checking ID")) {
    console.log("Adult");
}
```

`age >= 18` is false, so `console.log("Checking ID")` never runs.

### With `||`

If the left side is truthy, the right side is not evaluated.

```
let username = "Vinay";

if (username || "Guest") {
    console.log("A name exists");
}
```

Since `username` is truthy, JavaScript doesn't need the right-hand expression to determine the result.

Remember: `&&` and `||` return one of their operand values, not necessarily a Boolean. Inside an `if`, that value is then evaluated for truthiness.

## 7. Nested `if` vs. combined conditions

Both approaches can solve the same problem, but they express the logic differently.

Nested version:

```
let age = 20;
let hasID = true;

if (age >= 18) {
    if (hasID) {
        console.log("Allowed");
    }
}
```

Combined version:

```
if (age >= 18 && hasID) {
    console.log("Allowed");
}
```

Use the combined condition when all requirements simply need to be true. Use nested conditions when the second decision depends on the first, or when each level needs a distinct response.

## 8. The dangling `else` problem

An `else` belongs to the nearest unmatched `if`.

```
let age = 17;
let hasID = true;

if (age >= 18)
    if (hasID)
        console.log("Allowed");
    else
        console.log("ID required");
```

Here, `else` belongs to `if (hasID)`, not `if (age >= 18)`. Since the age condition is false, nothing prints.

Avoid this confusion by always using braces:

```
if (age >= 18) {
    if (hasID) {
        console.log("Allowed");
    } else {
        console.log("ID required");
    }
} else {
    console.log("Underage");
}
```

Best practice: Use `{}` even for one-line conditional blocks.

## 9. Common mistakes and edge cases

### Mistake 1: Using `=` instead of `===`

```
let age = 20;

if (age = 18) {
    console.log("This executes");
}
```

`age = 18` assigns the value `18` to `age`. The assignment expression evaluates to `18`, which is truthy.

Correct:

```
if (age === 18) {
    console.log("Exactly 18");
}
```

### Mistake 2: Incorrect range checking

Incorrect:

```
if (marks >= 40 || marks <= 100) {
    console.log("Valid marks");
}
```

This condition is true for almost every number because a number usually satisfies at least one side.

Correct:

```
if (marks >= 40 && marks <= 100) {
    console.log("Pass-range marks");
}
```

For validating the full range from 0 to 100:

```
if (marks >= 0 && marks <= 100) {
    console.log("Valid marks");
} else {
    console.log("Invalid marks");
}
```

### Mistake 3: Wrong condition order

```
let marks = 95;

if (marks >= 40) {
    console.log("Pass");
} else if (marks >= 90) {
    console.log("A grade");
}
```

This prints `Pass` because the first condition matches, and the remaining branches are skipped.

Correct:

```
if (marks >= 90) {
    console.log("A grade");
} else if (marks >= 40) {
    console.log("Pass");
} else {
    console.log("Fail");
}
```

When using `else if`, order conditions from the most specific or highest threshold to the more general ones.

### Mistake 4: Comparing floating-point numbers for exact equality

```
let result = 0.1 + 0.2;

if (result === 0.3) {
    console.log("Equal");
} else {
    console.log("Not exactly equal");
}
```

This typically prints `Not exactly equal` because of floating-point representation.

For approximate comparisons:

```
if (Math.abs(result - 0.3) < 0.000001) {
    console.log("Approximately equal");
}
```

## 10. Ternary operator: a compact alternative

For simple choices, JavaScript has a conditional operator `condition ? value1 : value2`.

```
let age = 20;

let message = age >= 18 ? "Adult" : "Minor";

console.log(message);
```

This is equivalent to:

```
let message;

if (age >= 18) {
    message = "Adult";
} else {
    message = "Minor";
}
```

Use ternary expressions for simple assignments. Avoid deeply nested ternaries because they can become difficult to read.

## 11. `if` vs. `switch`

- Use `if-else` when conditions involve ranges, comparisons, or multiple logical expressions.
- Use `switch` when comparing one expression against several specific values.

Example with `if-else`:

```
let marks = 82;

if (marks >= 90) {
    console.log("A");
} else if (marks >= 80) {
    console.log("B");
} else {
    console.log("Other grade");
}
```

Example with `switch`:

```
let day = 2;

switch (day) {
    case 1:
        console.log("Monday");
        break;
    case 2:
        console.log("Tuesday");
        break;
    default:
        console.log("Unknown day");
}
```

A `switch` case generally compares using strict equality. Remember `break` to prevent unintended fall-through.

## 12. Advanced best practices

1. Use `===` and `!==` for predictable comparisons.
2. Use braces `{}` for every `if`, `else if`, and `else` block.
3. Keep conditions readable; extract complicated logic into named Boolean variables or functions.
4. Check invalid inputs and boundary values, such as `0`, the minimum, the maximum, and just outside the range.
5. Avoid unnecessary nesting when a clear combined condition is sufficient.
6. Remember that `else if` branches are checked in order, and only the first matching branch runs.
7. Don't rely on truthiness when the difference between `0`, `""`, `null`, and `undefined` matters.
